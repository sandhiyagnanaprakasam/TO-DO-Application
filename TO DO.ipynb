```python

tasks = []
next_id = 1


def find_task(tid):
    for task in tasks:
        if task["id"] == tid:
            return task
    return None


def add_task():
    global next_id

    name = input("Enter task: ").strip()
    if not name:
        print("Task cannot be empty")
        return

    tasks.append({
        "id": next_id,
        "name": name,
        "status": "Pending"
    })

    print("Task added!")
    next_id += 1


def view_tasks():
    if not tasks:
        print("No tasks found")
        return

    for task in tasks:
        print(
            f"{task['id']} | {task['name']} | "
            f"{task['status']}"
        )


def get_task():
    try:
        tid = int(input("Enter task ID: "))
        task = find_task(tid)

        if task is None:
            print("Task not found")

        return task

    except ValueError:
        print("Invalid ID")
        return None


def update_task():
    task = get_task()
    if task:
        name = input("Enter new task: ").strip()
        if name:
            task["name"] = name
            print("Task updated!")


def complete_task():
    task = get_task()
    if task:
        task["status"] = "Completed"
        print("Task completed!")


def delete_task():
    task = get_task()
    if task:
        tasks.remove(task)
        print("Task deleted!")


def main():
    while True:
        print("\n1.Add 2.View 3.Update")
        print("4.Complete 5.Delete 6.Exit")

        choice = input("Choose: ")

        if choice == "1":
            add_task()
        elif choice == "2":
            view_tasks()
        elif choice == "3":
            update_task()
        elif choice == "4":
            complete_task()
        elif choice == "5":
            delete_task()
        elif choice == "6":
            break
        else:
            print("Invalid choice")


if __name__ == "__main__":
    main()
```
